```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/Churn_Modelling.csv")
```

```python
x = df.iloc[:,3:13]
y = df.iloc[:,13]
```

```python
from sklearn.preprocessing import LabelEncoder

labelencoder_x_1 = LabelEncoder()
x.iloc[:, 1] = labelencoder_x_1.fit_transform(x.iloc[:, 1])

labelencoder_x_2 = LabelEncoder()
x.iloc[:, 2] = labelencoder_x_2.fit_transform(x.iloc[:, 2])
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 0)
```

```python
len(df)
```

```python
len(x_test)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss = StandardScaler()
x_train = ss.fit_transform(x_train)
x_test = ss.transform(x_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
classifier = DecisionTreeClassifier()
classifier.fit(x_train, y_train)
pred = classifier.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
conM = confusion_matrix(y_test, pred)
print(conM)
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
print(accuracy_score(y_test, pred))
print(classification_report(y_test, pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rfc = RandomForestClassifier()
rfc.fit(x_train, y_train)
pred = rfc.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
conM = confusion_matrix(y_test, pred)
print(conM)
```

### Task 9: Improve the Existing ANN Model

To improve the model's accuracy by at least 5%, we will systematically explore different configurations, including:
-   **Hidden Layers and Neurons:** Further tuning the number of hidden layers and neurons per layer.
-   **Optimizer:** Fine-tuning the learning rate of existing optimizers or exploring new ones.
-   **Activation Functions:** Re-evaluating activation functions in conjunction with other changes.
-   **Dropout:** Adding dropout layers to prevent overfitting.
-   **Epochs:** Adjusting the number of training epochs.

We will start by defining a flexible function to build, compile, train, and evaluate the ANN model, making it easier to experiment with these changes.

```python
from keras.layers import Dropout

def build_and_evaluate_ann(num_hidden_layers, neurons_per_layer, optimizer_name='adam', learning_rate=0.001, activation_func='relu', dropout_rate=0.0, epochs=100, batch_size=10, kernel_initializer='uniform'):
    # Initialize the ANN
    classifier = Sequential()

    # Input layer
    classifier.add(Dense(units=neurons_per_layer, kernel_initializer=kernel_initializer, activation=activation_func, input_dim=x_train.shape[1]))
    if dropout_rate > 0.0:
        classifier.add(Dropout(dropout_rate))

    # Hidden layers
    for _ in range(num_hidden_layers - 1):
        classifier.add(Dense(units=neurons_per_layer, kernel_initializer=kernel_initializer, activation=activation_func))
        if dropout_rate > 0.0:
            classifier.add(Dropout(dropout_rate))

    # Output layer
    classifier.add(Dense(units=1, kernel_initializer='uniform', activation='sigmoid'))

    # Compile the ANN
    optimizer = None
    if optimizer_name == 'adam':
        optimizer = keras.optimizers.Adam(learning_rate=learning_rate)
    elif optimizer_name == 'sgd':
        optimizer = keras.optimizers.SGD(learning_rate=learning_rate)
    elif optimizer_name == 'rmsprop':
        optimizer = keras.optimizers.RMSprop(learning_rate=learning_rate)
    elif optimizer_name == 'adagrad':
        optimizer = keras.optimizers.Adagrad(learning_rate=learning_rate)
    elif optimizer_name == 'nadam':
        optimizer = keras.optimizers.Nadam(learning_rate=learning_rate)
    else:
        raise ValueError(f"Unknown optimizer: {optimizer_name}")

    classifier.compile(optimizer=optimizer, loss='binary_crossentropy', metrics=['accuracy'])

    # Train the ANN
    print(f"\nTraining ANN with {num_hidden_layers} hidden layers, {neurons_per_layer} neurons, {optimizer_name} (lr={learning_rate}), {activation_func} activation, dropout={dropout_rate}, epochs={epochs}, batch_size={batch_size}")
    history = classifier.fit(x_train, y_train, batch_size=batch_size, epochs=epochs, verbose=0)

    # Predict and evaluate
    y_pred_ann = classifier.predict(x_test)
    y_pred_ann = (y_pred_ann > 0.5)
    cm = confusion_matrix(y_test, y_pred_ann)
    accuracy = accuracy_score(y_test, y_pred_ann)
    final_loss = history.history['loss'][-1]

    print(f"Confusion Matrix:\n{cm}")
    print(f"Test Accuracy: {accuracy:.4f}")
    print(f"Last Epoch Training Loss: {final_loss:.4f}")

    return accuracy, final_loss
```
